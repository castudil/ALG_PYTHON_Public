## Parte 1 — El Fin del Tipado Estricto

### ¿Qué cambió?

En Java, el tipo de una variable es **fijo en tiempo de compilación**. El compilador es el guardián:

```java
// Java: el tipo está pegado a la variable para siempre
int contador = 10;
contador = "hola"; // ERROR de compilación
```

En Python, el tipo vive en el **objeto**, no en la variable. La variable es solo una etiqueta que puede apuntar a cualquier objeto:

```
contador ──▶ [objeto int: 10]
contador ──▶ [objeto str: "hola"]   ← perfectamente válido
```

Esto se llama **tipado dinámico**: el tipo se resuelve en *tiempo de ejecución*.  
No es "tipado débil" — Python es en realidad muy estricto sobre las *operaciones* permitidas entre tipos, pero flexible sobre a qué tipo apunta una variable.

| Concepto | Java | Python |
|---|---|---|
| Tipo asociado a | La **variable** | El **objeto** |
| Verificación | Compilación | Ejecución |
| Cambiar tipo en variable | ❌ Error | ✅ Permitido |
| Mezclar tipos en colección | ❌ (sin generics) | ✅ Nativo |

````python
# === Tipado Dinámico en Acción ===
# La función type() nos muestra el tipo del objeto al que apunta la variable.

x = 42
print(f"x = {x!r}  →  tipo: {type(x).__name__}")

x = "ahora soy texto"
print(f"x = {x!r}  →  tipo: {type(x).__name__}")

x = 3.14
print(f"x = {x!r}  →  tipo: {type(x).__name__}")

x = [1, "dos", 3.0, True]  # lista heterogénea — impensable en Java sin Object[]
print(f"x = {x!r}  →  tipo: {type(x).__name__}")
````

> 🎙️ **[PAUSA PROFESOR]** Pregunta sugerida: *"¿Qué imprime `a = [1,2,3]; b = a; b.append(4); print(a)`? ¿Por qué?"* — Es la puerta de entrada a la diferencia entre nombre y objeto, y explica el bug de matrices de la Parte 5.

> **Nota para ex-JavaDevs:** `print(f"...")` es un *f-string*, el equivalente moderno de `String.format()`. La `!r` aplica `repr()` al valor, mostrando el tipo con comillas para strings, como lo haría un debugger.

## Parte 2 — El Fin del `while` Forzado: Bienvenido el `for` Idiomático

### El patrón que aprendiste (y por qué lo debes olvidar para iterar)

En tu curso anterior de Python te enseñaron esto:

```python
# Estilo "Java disfrazado de Python" — EVITAR para iteración simple
i = 0
while i < len(mi_lista):
    print(mi_lista[i])
    i += 1
```

En Java el `for` clásico también trabaja con un índice:

```java
// Java
for (int i = 0; i < lista.size(); i++) {
    System.out.println(lista.get(i));
}
```

Ambos son válidos, pero en Python tenemos algo más poderoso: el **protocolo de iteración**.  
Python puede iterar sobre *cualquier objeto que sepa entregarte elementos uno a uno*. No necesitas saber cuántos hay ni llevar un contador:

```python
# Python idiomático
for elemento in coleccion:
    print(elemento)
```

### ¿Cuándo SÍ necesitas el índice? → `enumerate()`

Si necesitas tanto el índice como el valor (como en Java), Python no te pide que gestiones el contador: te da `enumerate()`.

```java
// Java
for (int i = 0; i < arr.length; i++) {
    System.out.println(i + ": " + arr[i]);
}
```
```python
# Python idiomático equivalente
for i, valor in enumerate(arr):
    print(f"{i}: {valor}")
```

````python
# === Comparativa de patrones de iteración ===

frutas: list[str] = ["manzana", "banana", "cereza", "durazno"]

print("--- Estilo while con índice (evitar) ---")
i = 0
while i < len(frutas):
    print(f"  {frutas[i]}")
    i += 1

print("\n--- Estilo for idiomático (preferido) ---")
for x in frutas:
    print(f"  {x}")

print("\n--- Con índice usando enumerate() ---")
for indice, fruta in enumerate(frutas):
    print(f"  [{indice}] {fruta}")
````

## Parte 3 — `range()`: Cuando Sí Necesitas un Contador

`range()` genera una secuencia de enteros *bajo demanda* (es un objeto perezoso, no crea una lista en memoria). Es el puente entre el `for(int i=0; i<n; i++)` de Java y el `for` idiomático de Python.

| Forma | Equivalente Java | Descripción |
|---|---|---|
| `range(n)` | `i = 0; i < n; i++` | De 0 a n-1 |
| `range(a, b)` | `i = a; i < b; i++` | De a a b-1 |
| `range(a, b, paso)` | `i = a; i < b; i += paso` | Con paso personalizado |

````python
# === Variantes de range() ===

print("range(5)       →", list(range(5)))
print("range(2, 8)    →", list(range(2, 8)))
print("range(0, 10, 2)→", list(range(0, 10, 2)))
print("range(10, 0,-2)→", list(range(10, 0, -2)))  # ¡cuenta regresiva!

print("\n--- range() como sustituto del for(int i) de Java ---")
n: int = 5
for i in range(n):
    print(f"  iteración {i}")

print("\n--- Construir una tabla de multiplicar con range() ---")
base: int = 7
for i in range(1, 11):
    print(f"  {base} × {i:2d} = {base * i:3d}")
````

## Parte 4 — Listas: Creación Dinámica vs. Pre-asignación

### El modelo mental de Java

En Java un array tiene tamaño fijo definido en la declaración:
```java
int[] notas = new int[30];  // reserva 30 espacios, todos en 0
```

### Python: listas son dinámicas por naturaleza

Una lista Python puede crecer o decrecer en tiempo de ejecución. No hay tamaño fijo. Sin embargo, a veces es útil pre-crear una lista con valores iniciales (por ejemplo, para llenar posiciones por índice).

| Situación | Patrón Python |
|---|---|
| Acumular resultados uno a uno | `lista = []` + `lista.append(x)` |
| Pre-crear con N ceros | `lista = [0] * n` |
| Pre-crear con valores calculados | Comprensión de lista |
| Matriz 2D correcta | Comprensión de lista anidada |

````python
# === Creación dinámica vs. pre-asignación ===

# Patrón 1: lista vacía + append (equivale a un ArrayList en Java)
cuadrados: list[int] = []
for i in range(1, 6):
    cuadrados.append(i ** 2)
print("Cuadrados (append):", cuadrados)

# Patrón 2: pre-asignación con operador de repetición
# Útil cuando sabemos el tamaño y llenaremos por índice después
n: int = 5 # tamaño del buffer, equivalente a new int[n] en Java
buffer: list[int] = [0] * n # pre-asigna un buffer de 5 enteros inicializados a 0
print("Buffer pre-asignado:", buffer) # muestra [0, 0, 0, 0, 0]

# Llenar el buffer por índice (patrón válido)
for i in range(n): # i va de 0 a 4, índices válidos para buffer
    buffer[i] = (i + 1) * 10 # asigna 10, 20, 30, 40, 50 a las posiciones del buffer
print("Buffer lleno por índice:", buffer) # muestra [10, 20, 30, 40, 50]

# Patrón 3: comprensión de lista (el más idiomático para transformaciones)
cuadrados_comp: list[int] = [i ** 2 for i in range(1, 6)] # construye la lista de cuadrados directamente. funciona como un map + filter en una sola expresión. es la forma más "pythónica" de crear listas transformadas. primero se evalúa el for i in range(1, 6) para generar los valores de i, luego se calcula i ** 2 para cada i y se construye la lista resultante. es equivalente a map(lambda i: i**2, range(1, 6)) pero más legible y flexible.
print("Cuadrados (comprensión):", cuadrados_comp)
````

## Parte 5 — El Bug Clásico: Matrices por Referencia

> **Este es uno de los errores más comunes al venir de Java.** Presta atención.

### El intento "lógico" (y equivocado)

Cuando queremos una matriz de `filas × columnas` inicializada en cero, la primera intuición es usar el operador `*` que ya aprendimos:

```python
matriz = [[0] * columnas] * filas  # ← TRAMPA
```

### ¿Por qué falla?

El operador `*` en listas **copia la referencia**, no el objeto.  
Esto significa que las `filas` referencias apuntan **al mismo objeto fila** en memoria:

```
matriz[0] ──▶ [fila_A]
matriz[1] ──▶ [fila_A]  ← ¡mismo objeto!
matriz[2] ──▶ [fila_A]  ← ¡mismo objeto!
```

Modificar `matriz[0][1]` también modifica `matriz[1][1]` y `matriz[2][1]` — todas apuntan a la misma lista.

En Java esto no ocurre porque `new int[filas][cols]` reserva memoria independiente para cada fila. Python, al ser un lenguaje de referencias, requiere que creemos cada fila de forma independiente.

````python
# === DEMOSTRACIÓN DEL BUG: [[0] * m] * n ===

filas: int = 3
cols: int = 4

# --- Forma INCORRECTA ---
matriz_bug: list[list[int]] = [[0] * cols] * filas

print("Antes de modificar:")
for fila in matriz_bug:
    print(" ", fila)

# Intentamos modificar solo la celda [0][1]
matriz_bug[0][1] = 99

print("\nDespués de modificar SOLO matriz_bug[0][1] = 99:")
for fila in matriz_bug:
    print(" ", fila)

print("\n¡Las 3 filas cambiaron! Todas apuntan al mismo objeto en memoria.")

# Confirmamos que son el mismo objeto con id()
print(f"\nid(matriz_bug[0]) = {id(matriz_bug[0])}")
print(f"id(matriz_bug[1]) = {id(matriz_bug[1])}")
print(f"id(matriz_bug[2]) = {id(matriz_bug[2])}")
print("¿Son iguales?", id(matriz_bug[0]) == id(matriz_bug[1]))
````

````python
# === FORMA CORRECTA: Comprensión de lista anidada ===
# Cada iteración del for crea una NUEVA lista independiente.

filas: int = 3
cols: int = 4

# La comprensión crea un objeto [0]*cols nuevo en cada iteración
matriz_ok: list[list[int]] = [[0] * cols for _ in range(filas)]
# Nota: usamos _ como nombre de variable cuando no necesitamos el valor del índice

print("Antes de modificar:")
for fila in matriz_ok:
    print(" ", fila)

# Modificamos SOLO la celda [0][1]
matriz_ok[0][1] = 99

print("\nDespués de modificar SOLO matriz_ok[0][1] = 99:")
for fila in matriz_ok:
    print(" ", fila)

print("\nSolo la fila 0 cambió.")

# Confirmamos que son objetos diferentes
print(f"\nid(matriz_ok[0]) = {id(matriz_ok[0])}")
print(f"id(matriz_ok[1]) = {id(matriz_ok[1])}")
print(f"id(matriz_ok[2]) = {id(matriz_ok[2])}")
print("¿Son iguales?", id(matriz_ok[0]) == id(matriz_ok[1]))
````

## Parte 6 — Iteración Idiomática sobre Matrices

Ahora que tenemos una matriz bien construida, veamos cómo recorrerla de forma idiomática.  
Compara con el doble `for` anidado con índices que usarías en Java:

```java
// Java
for (int i = 0; i < filas; i++) {
    for (int j = 0; j < cols; j++) {
        System.out.print(matriz[i][j] + " ");
    }
}
```

````python
# === Construir y recorrer una matriz con valores reales ===
from typing import Final

FILAS: Final[int] = 3
COLS: Final[int] = 4

# Construimos una matriz donde cada celda vale fila * col
# (tabla de multiplicar truncada)
tabla: list[list[int]] = [
    [(f + 1) * (c + 1) for c in range(COLS)]
    for f in range(FILAS)
]

print("=== Matriz (tabla de multiplicar) ===")

# Opción A: for idiomático (sin índices, cuando no los necesitas)
print("\nOpción A — for idiomático:")
for fila in tabla:
    print(" ", fila)

# Opción B: con enumerate para tener el índice de fila y columna
print("\nOpción B — con enumerate (acceso a índices):")
for i, fila in enumerate(tabla):
    for j, valor in enumerate(fila):
        print(f"  tabla[{i}][{j}] = {valor}")

# Opción C: impresión formateada como cuadrícula
print("\nOpción C — visualización como cuadrícula:")
for fila in tabla:
    print("  " + "  ".join(f"{v:3d}" for v in fila))
````

## Resumen Visual

| Concepto | Java | Python idiomático |
|---|---|---|
| Tipo en variable | `int x = 5;` | `x = 5` (dinámico) |
| Iterar colección | `for(int i=0; i<n; i++)` | `for elem in coleccion:` |
| Iterar con índice | `for(int i=0; ...)` | `for i, e in enumerate(c):` |
| Secuencia numérica | `for(int i=0; i<n; i++)` | `for i in range(n):` |
| Lista vacía crecer | `ArrayList<>()` + `add()` | `[]` + `append()` |
| Array de N ceros | `new int[n]` | `[0] * n` |
| Matriz 2D correcta | `new int[f][c]` | `[[0]*c for _ in range(f)]` |
| Matriz 2D TRAMPA | — | `[[0]*c] * f` ← ¡NO! |
