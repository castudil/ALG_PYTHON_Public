### Bloque 1 y 2: Análisis de la Implementación Base
Ejecuta la siguiente celda y analiza la estructura de datos y el algoritmo de búsqueda lineal. 
* Complejidad teórica: Tiempo $\mathcal{O}(n)$, Espacio $\mathcal{O}(1)$.

```python
def busqueda_maximo(arreglo):
    if not arreglo:
        return None
    
    maximo = arreglo[0]
    for elemento in arreglo:
        if elemento > maximo:
            maximo = elemento
            
    return maximo

temperaturas = [15.5, 18.2, 14.1, 22.0, 19.3]
print(f"Temperatura máxima detectada: {busqueda_maximo(temperaturas)}")
```

```python
# Pruebas del Bloque 3 con unittest — NO MODIFICAR
import unittest


class ListaVigilada(list):
    """Lista que cuenta cuántos elementos se leen (por iteración o por índice)."""

    def __init__(self, datos):
        super().__init__(datos)
        self.leidos = 0

    def __iter__(self):
        for x in super().__iter__():
            self.leidos += 1
            yield x

    def __getitem__(self, i):
        self.leidos += 1
        return super().__getitem__(i)


class TestTaxonomia(unittest.TestCase):
    """Problemas de decisión, conteo y optimización sobre una lista de temperaturas."""

    def test_1_decision(self):
        """existe_temperatura_critica: True solo si alguna temperatura supera el umbral"""
        self.assertIs(existe_temperatura_critica(temperaturas, 20.0), True)
        self.assertIs(existe_temperatura_critica(temperaturas, 22.0), False)   # 22.0 no es superior a 22.0
        self.assertIs(existe_temperatura_critica([], 0.0), False)

    def test_2_decision_cortocircuito(self):
        """existe_temperatura_critica deja de leer al encontrar el primer caso"""
        datos = ListaVigilada([35.0] + [10.0] * 999)
        self.assertIs(existe_temperatura_critica(datos, 30.0), True)
        self.assertLessEqual(datos.leidos, 2, "siguió recorriendo tras encontrar el primer caso")

    def test_3_conteo(self):
        """contar_temperaturas_bajas cuenta las temperaturas estrictamente menores al umbral"""
        self.assertEqual(contar_temperaturas_bajas(temperaturas, 16.0), 2)
        self.assertEqual(contar_temperaturas_bajas(temperaturas, 14.1), 0)
        self.assertEqual(contar_temperaturas_bajas(temperaturas, 100.0), 5)
        self.assertEqual(contar_temperaturas_bajas([], 16.0), 0)

    def test_4_optimizacion(self):
        """optimizacion_min_max retorna la tupla (mínimo, máximo)"""
        self.assertEqual(optimizacion_min_max(temperaturas), (14.1, 22.0))
        self.assertEqual(optimizacion_min_max([7.5]), (7.5, 7.5))
        self.assertEqual(optimizacion_min_max([-3.0, 5.0, -8.0]), (-8.0, 5.0))

    def test_5_optimizacion_una_pasada(self):
        """optimizacion_min_max lee el arreglo una sola vez"""
        datos = ListaVigilada([float(x) for x in range(100)])
        optimizacion_min_max(datos)
        self.assertLessEqual(datos.leidos, len(datos) + 2, "leyó el arreglo más de una vez")


resultado = unittest.main(argv=["ignorado", "TestTaxonomia"], exit=False, verbosity=2)
```

### Bloque 4: Verificación Empírica de Complejidad
Comprobación del modelo asintótico lineal $\mathcal{O}(n)$ mediante la inyección de arreglos de gran magnitud. Ejecuta la celda y analiza el output.

```python
import random
import time

def medir_tiempo_ejecucion(tamano_n):
    # Generación programática del arreglo en memoria
    datos_prueba = [random.uniform(0.0, 40.0) for _ in range(tamano_n)]
    
    inicio = time.perf_counter()
    _ = busqueda_maximo(datos_prueba)
    fin = time.perf_counter()
    
    tiempo_transcurrido = fin - inicio
    return tiempo_transcurrido

# Prueba asintótica
n_base = 1_000_000
n_escala = 10_000_000

print("Iniciando evaluación empírica...")
tiempo_base = medir_tiempo_ejecucion(n_base)
print(f"Tiempo para n={n_base:,}: {tiempo_base:.4f} segundos")

tiempo_escala = medir_tiempo_ejecucion(n_escala)
print(f"Tiempo para n={n_escala:,}: {tiempo_escala:.4f} segundos")

incremento_n = n_escala / n_base
incremento_tiempo = tiempo_escala / tiempo_base
print(f"\nAnálisis: Al aumentar la entrada en {incremento_n}x, el tiempo aumentó en {incremento_tiempo:.2f}x. Esto comprueba empíricamente O(n).")
```
