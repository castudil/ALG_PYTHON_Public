````python
import math
import sys
import unittest

def mostrar_barras(filas, unidad="ms", ancho=40, formato="8.2f", maximo=None):
    """
    Imprime un gráfico de barras horizontal en texto.

    Parámetros:
        filas (list[tuple[str, float]]): pares (etiqueta, valor).
        unidad (str): unidad que se muestra junto a cada valor.
        ancho (int): caracteres de la barra que representa a `maximo`.
        formato (str): formato del valor impreso, por ejemplo "8.2f" o "8,.0f".
        maximo (float | None): valor que ocupa todo el ancho; por defecto, el mayor
            de `filas`. Pasar el mismo valor a dos gráficos los deja en la misma escala.
    """
    maximo = maximo or max(valor for _, valor in filas) or 1
    margen = max(len(etiqueta) for etiqueta, _ in filas)
    for etiqueta, valor in filas:
        barra = "█" * round(valor / maximo * ancho) or ("▏" if valor > 0 else "")
        print(f"{etiqueta:>{margen}} │{barra:<{ancho}} {valor:{formato}} {unidad}")


print("🐍 Python", sys.version.split()[0], "| Todo listo para comenzar.")
````

### 2. Taxonomía de Problemas Computacionales
Clasificación estructurada bajo la Teoría de la Computabilidad y Complejidad.

**Según la Naturaleza del Resultado:**
* **De Decisión:** Respuesta binaria (Sí/No).
* **De Búsqueda:** Localización de una estructura o valor.
* **De Optimización:** Búsqueda de la mejor solución (minimizar/maximizar).
* **De Conteo:** Cálculo de la cardinalidad de soluciones válidas.

**Según Computabilidad:**
* **Decidibles:** Existe un algoritmo que siempre se detiene.
* **Indecidibles:** No existe un algoritmo general (ej. Problema de la Parada).

**Según Complejidad Computacional:**
* **P:** Resolubles en tiempo polinómico determinista $\mathcal{O}(n^k)$.
* **NP:** Soluciones verificables en tiempo polinómico determinista.
* **NP-Completo:** Los problemas más difíciles de NP.
* **NP-Hard:** Al menos tan difíciles como los NP-Completos, no restringidos a decisión.

### 7. Contar operaciones: de los segundos a $T(n)$

El tiempo en segundos depende del computador, del lenguaje y de qué más esté corriendo. Para
comparar algoritmos necesitamos una medida que dependa **solo del algoritmo**: el número de
operaciones básicas que ejecuta, como función del tamaño de la entrada $n$. A esa función la
llamamos $T(n)$.

**Ejemplo 1 — buscar el máximo.** El ciclo de la celda anterior compara cada elemento una vez:

$$T(n) = n \quad \text{comparaciones}$$

**Ejemplo 2 — dos ciclos anidados.**

```python
for i in range(n):
    for j in range(i, n):
        operacion()
```

Para `i = 0` el ciclo interno hace $n$ vueltas; para `i = 1`, $n-1$; … para `i = n-1`, una sola:

$$T(n) = n + (n-1) + \dots + 1 = \frac{n(n+1)}{2}$$

No hace falta ejecutar el código para saberlo, pero podemos comprobarlo contando.


````python
# Contamos las operaciones de verdad y las comparamos con la fórmula
def contar_maximo(n):
    """Comparaciones que hace la búsqueda del máximo sobre n elementos."""
    datos = list(range(n))
    comparaciones = 0
    maximo = datos[0] if n > 0 else None
    for x in datos:
        comparaciones += 1
        if x > maximo:
            maximo = x
    return comparaciones


def contar_anidado(n):
    """Veces que se ejecuta el cuerpo de los dos ciclos anidados."""
    veces = 0
    for i in range(n):
        for j in range(i, n):
            veces += 1
    return veces


print(f"{'n':>6} │ {'máximo':>8} {'T(n)=n':>8} │ {'anidado':>10} {'n(n+1)/2':>10}")
print("─" * 50)
for n in (1, 10, 100, 1000):
    print(f"{n:>6} │ {contar_maximo(n):>8,} {n:>8,} │ {contar_anidado(n):>10,} {n * (n + 1) // 2:>10,}")

print("\n💡 Al multiplicar n por 10, el primero crece ×10 y el segundo ×100:")
print("   eso es lo que la notación asintótica va a capturar.")
````

### 8. Notación asintótica: $O$, $\Omega$ y $\Theta$

De $T(n) = \frac{n(n+1)}{2} = \frac{1}{2}n^2 + \frac{1}{2}n$ nos interesa **cómo crece**, no
el valor exacto. Para $n$ grande manda el término $n^2$, y la constante $\frac{1}{2}$ depende
de qué contamos como «una operación». La notación asintótica se queda con lo esencial:

| Notación | Se lee | Significa | Para $T(n) = \frac{n(n+1)}{2}$ |
|----------|--------|-----------|-------------------------------|
| $T(n) = O(g(n))$ | «O grande» | $T$ crece **a lo más** como $g$: cota superior | $O(n^2)$ ✓ · también $O(n^3)$ ✓ (cierto, pero dice poco) |
| $T(n) = \Omega(g(n))$ | «omega» | $T$ crece **al menos** como $g$: cota inferior | $\Omega(n^2)$ ✓ · también $\Omega(n)$ ✓ |
| $T(n) = \Theta(g(n))$ | «theta» | $T$ crece **exactamente** como $g$: cota ajustada | $\Theta(n^2)$ ✓ · $\Theta(n^3)$ ✗ |

Formalmente, $T(n) = O(g(n))$ si existen constantes $c > 0$ y $n_0$ tales que
$T(n) \le c \cdot g(n)$ para todo $n \ge n_0$. Para $\Omega$ la desigualdad es $\ge$, y
$\Theta$ exige las dos a la vez.


> 🎙️ **[PAUSA PROFESOR]** Pregunta sugerida: *"¿Es correcto decir que la búsqueda binaria es $O(n^2)$?"* — Sí lo es, y por eso $O$ sola dice poco: es una cota superior, no una descripción exacta. Ahí entra $\Theta$.

### 9. Peor caso, mejor caso y caso promedio

Un mismo algoritmo puede costar distinto según **cuál** sea la entrada, aunque el tamaño sea el
mismo. La búsqueda lineal de un valor en un arreglo de $n$ elementos:

| Caso | ¿Qué entrada? | Comparaciones |
|------|---------------|---------------|
| **Mejor** | El valor está en la primera posición | 1 |
| **Peor** | El valor está al final, o no está | $n$ |
| **Promedio** | El valor está en una posición al azar | $\approx n/2$ |

> ⚠️ **No confundir casos con cotas.** «Peor caso» y «mejor caso» hablan de la **entrada**;
> $O$, $\Omega$ y $\Theta$ hablan de **cómo crece** una función. Son cosas independientes: el peor
> caso de la búsqueda lineal es $\Theta(n)$, y su mejor caso es $\Theta(1)$. Cuando se dice
> «este algoritmo es $O(n)$» sin más, se suele entender que es una cota para el peor caso.

Casi siempre nos interesa el **peor caso**: es una garantía. El caso promedio importa cuando el
peor es raro (como veremos con Quicksort).


````python
# Los tres casos de la búsqueda lineal, contados
def busqueda_lineal_contada(datos, objetivo):
    """Retorna (índice o -1, comparaciones realizadas)."""
    comparaciones = 0
    for i in range(len(datos)):
        comparaciones += 1
        if datos[i] == objetivo:
            return i, comparaciones
    return -1, comparaciones


n = 1000
datos = list(range(n))
mejor = busqueda_lineal_contada(datos, 0)[1]
peor = busqueda_lineal_contada(datos, -1)[1]                     # no está
promedio = sum(busqueda_lineal_contada(datos, x)[1] for x in datos) / n

print(f"Búsqueda lineal sobre n = {n} elementos\n")
mostrar_barras([("mejor caso", mejor), ("caso promedio", promedio), ("peor caso", peor)],
               unidad="comparaciones", formato="7.1f")
````

### 10. Órdenes de crecimiento típicos

Estos cinco órdenes aparecerán una y otra vez en el curso. La tabla muestra cuántas operaciones
representa cada uno; si un computador hace unas $10^8$ operaciones por segundo, todo lo que pase
de $10^{10}$ ya son minutos.

> 🎙️ **[PAUSA PROFESOR]** *"Antes de ejecutar la celda: para n = 1.000.000, ¿cuál de estos órdenes sigue siendo viable?"*
> *$\log n$, $n$ y $n \log n$ sin problemas (hasta 2 × 10⁷ operaciones). $n^2$ son 10¹² operaciones: horas. $2^n$ ya era inviable con n = 100.*


````python
# Cuánto vale cada orden de crecimiento
def legible(x):
    """Formatea un número grande: con separadores si es chico, en potencias de 10 si no."""
    if x < 1e15:
        return f"{x:,.0f}"
    return f"≈ 10^{math.log10(x):.0f}"


print(f"{'n':>10} │ {'log₂ n':>8} {'n':>12} {'n log₂ n':>14} {'n²':>18} {'2ⁿ':>14}")
print("─" * 82)
for n in (10, 100, 1000, 1_000_000):
    exponente = n * math.log10(2)
    dos_n = f"{2 ** n:,}" if n <= 40 else f"≈ 10^{exponente:,.0f}"
    print(f"{n:>10,} │ {math.log2(n):>8.1f} {n:>12,} {legible(n * math.log2(n)):>14} "
          f"{legible(n ** 2):>18} {dos_n:>14}")

print("\nPara n = 20, en la misma escala:\n")
mostrar_barras([("log₂ n", math.log2(20)), ("n", 20), ("n log₂ n", 20 * math.log2(20)),
                ("n²", 20 ** 2), ("2ⁿ", 2 ** 20)], unidad="", formato="11,.0f")
print("\n💡 2ⁿ aplasta a todos los demás ya con n = 20. Con n = 100 son ≈ 10^30 operaciones:")
print("   a 10^8 por segundo, mucho más que la edad del universo.")
````
