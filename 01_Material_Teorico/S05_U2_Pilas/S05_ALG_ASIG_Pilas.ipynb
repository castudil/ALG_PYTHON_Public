## Descripción del Problema: El Navegador de la Nave Espacial

### Contexto
El sistema de navegación de una nave espacial tiene un navegador de URLs que permite moverse hacia atrás y adelante en el historial de páginas visitadas, similar al navegador de tu computadora.

### Comandos del Sistema
- **VISITAR "url"** — visita una nueva URL (borra el historial de "adelante")
- **ATRAS** — vuelve a la página anterior (si existe)
- **ADELANTE** — avanza a la siguiente página (si existe)
- **ACTUAL** — retorna la URL actual

### Casos Límite Críticos
1. **ATRAS cuando estás en la primera página** → no hace nada (sin error), retorna URL actual
2. **ADELANTE cuando no hay páginas adelante** → no hace nada, retorna URL actual
3. **VISITAR nueva página cuando hay historial de adelante** → **el historial de adelante se borra**
4. **VISITAR cuando es la primera visita** → pila de atrás está vacía

### Estructura Interna: Dos Pilas
```
┌──────────────────────────────────────────────────┐
│                                                    │
│  pila_atras        actual          pila_adelante  │
│  [LIFO]            [str]           [LIFO]         │
│                                                    │
│  base.com      google.com          nasa.gov       │
│  google.com    ←────────→          ────────→      │
│                                                    │
│  La URL actual NO está en ninguna pila.           │
│  pila_atras almacena el historial de atrás.       │
│  pila_adelante almacena el historial de adelante. │
└──────────────────────────────────────────────────┘
```

### Ejemplo de Sesión Completa
```
VISITAR "base.com"
  actual: "base.com",   atras: [],                  adelante: []

VISITAR "google.com"
  actual: "google.com",  atras: [base.com],         adelante: []

VISITAR "nasa.gov"
  actual: "nasa.gov",    atras: [base.com, google.com],  adelante: []
                          (tope = google.com)

ATRAS
  actual: "google.com",  atras: [base.com],         adelante: [nasa.gov]

ATRAS
  actual: "base.com",    atras: [],                 adelante: [nasa.gov, google.com]
                                                      (tope = google.com)

ADELANTE
  actual: "google.com",  atras: [base.com],         adelante: [nasa.gov]

VISITAR "esa.int"
  actual: "esa.int",     atras: [base.com, google.com],  adelante: []  ← ¡Se borró nasa.gov!

ATRAS
  actual: "google.com",  atras: [base.com],         adelante: [esa.int]
```

## API de la Librería `pilas.py`

| Método | Firma | Descripción | Costo |
|--------|-------|-------------|-------|
| **push** | `push(item) → None` | Inserta un elemento en el tope de la pila | O(1) amortizado |
| **pop** | `pop() → Any` | Elimina y retorna el elemento del tope | O(1) amortizado |
| **peek** | `peek() → Any` | Ve el tope sin eliminar | O(1) |
| **is_empty** | `is_empty() → bool` | Retorna True si la pila está vacía | O(1) |
| **size** | `size() → int` | Retorna el número de elementos | O(1) |
| **len** | `len(p) → int` | Equivalente a `p.size()` | O(1) |
| **repr** | `repr(p) → str` | Representación textual para debugging | O(n) |

### Errores Importantes
⚠️ **`pop()` y `peek()` lanzan `IndexError` si la pila está vacía.**  
**Siempre verifica `is_empty()` antes de llamarlos cuando puedan estar vacíos.**

## Representación Interna: Mapeando Dominio → Estructura

```
Dominio (lo que ves):           Estructura Interna (lo que usas):
─────────────────────────────   ──────────────────────────────────
URL actual: "google.com"        self.__actual = "google.com" (str)
Historial atrás: ["base.com"]   self.__pila_atras = PilaArreglo()
Historial adelante: ["nasa.gov"] self.__pila_adelante = PilaArreglo()
```

### Lógica de Operaciones

#### 1. Al presionar ATRAS:
```python
if pila_atras no está vacía:
    1. push(actual) → pila_adelante
    2. actual = pop(pila_atras)
sino:
    no hacer nada, retornar actual
```

#### 2. Al presionar ADELANTE:
```python
if pila_adelante no está vacía:
    1. push(actual) → pila_atras
    2. actual = pop(pila_adelante)
sino:
    no hacer nada, retornar actual
```

#### 3. Al VISITAR nueva URL:
```python
1. if actual != None:
       push(actual) → pila_atras
2. vaciar pila_adelante (while not pila_adelante.is_empty(): pila_adelante.pop())
3. actual = nueva_url
```

**Punto clave**: Vaciar pila_adelante es fundamental. Sin esto, si navegas atrás y luego visitas una nueva URL, el navegador "recordaría" URLs que ya no son accesibles lógicamente.
