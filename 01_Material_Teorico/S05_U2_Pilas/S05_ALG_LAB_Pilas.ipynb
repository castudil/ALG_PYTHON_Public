Comparamos el desempeño de `PilaArreglo` vs `MiPilaEnlazada` en 10,000 operaciones push + pop.

**Pregunta clave:** ¿Cuál es más rápida? ¿Por qué?

````python
# ═══ NO MODIFICAR ═══
# Benchmark: PilaArreglo vs MiPilaEnlazada

import time

N = 10_000

def benchmark(clase_pila, nombre: str, n: int) -> float:
    p = clase_pila()
    inicio = time.perf_counter()
    for i in range(n):
        p.push(i)
    for _ in range(n):
        p.pop()
    fin = time.perf_counter()
    ms = (fin - inicio) * 1000
    print(f"  {nombre:25s}: {ms:.3f} ms para {n:,} push + {n:,} pop")
    return ms

print("─" * 55)
print("BENCHMARK: PilaArreglo vs MiPilaEnlazada")
print("─" * 55)
t1 = benchmark(PilaArreglo, "PilaArreglo (referencia)", N)
t2 = benchmark(MiPilaEnlazada, "MiPilaEnlazada (tuya)", N)
ratio = t2 / t1 if t1 > 0 else 0
print(f"\n  Ratio MiPilaEnlazada / PilaArreglo: {ratio:.2f}x")
print("─" * 55)
````
