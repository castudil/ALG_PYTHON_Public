## PARTE 4: Benchmark - Comparación de Rendimiento (5 minutos)

Compara el rendimiento de **FilaDeque** (basada en `collections.deque`) vs **MiFilaEnlazada** con 10,000 operaciones.

````python
# NO MODIFICAR ESTA CELDA

import time

def benchmark_filas(n=10000):
    """Compara el rendimiento de FilaDeque vs MiFilaEnlazada"""
    
    print(f"Benchmark: {n:,} operaciones enqueue + dequeue\n")
    print("="*60)
    
    # Benchmark FilaDeque
    print("\n1. FilaDeque (basada en collections.deque)...")
    start = time.time()
    f_deque = FilaDeque()
    for i in range(n):
        f_deque.enqueue(i)
    for i in range(n):
        f_deque.dequeue()
    time_deque = time.time() - start
    print(f"   Tiempo: {time_deque:.4f} segundos")
    
    # Benchmark MiFilaEnlazada
    print("\n2. MiFilaEnlazada (lista enlazada)...")
    start = time.time()
    f_enlazada = MiFilaEnlazada()
    for i in range(n):
        f_enlazada.enqueue(i)
    for i in range(n):
        f_enlazada.dequeue()
    time_enlazada = time.time() - start
    print(f"   Tiempo: {time_enlazada:.4f} segundos")
    
    # Análisis
    print("\n" + "="*60)
    ratio = time_enlazada / time_deque
    print(f"\nRatio (MiFilaEnlazada / FilaDeque): {ratio:.2f}x")
    
    if ratio > 1:
        print(f"→ MiFilaEnlazada es {ratio:.2f}x MÁS LENTA")
    else:
        print(f"→ MiFilaEnlazada es {1/ratio:.2f}x MÁS RÁPIDA")
    
    print(f"\nGanancia absoluta: {time_enlazada - time_deque:.4f} segundos")
    print("\n" + "="*60)

benchmark_filas()
````
