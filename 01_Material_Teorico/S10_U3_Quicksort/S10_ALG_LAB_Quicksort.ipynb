## PARTE 2B: Implementar la partición tricotómica de Dijkstra (20 minutos)

Cuando hay **muchas claves repetidas**, la partición de dos punteros reprocesa una y otra vez
elementos que ya están en su lugar. La partición tricotómica los deja fijos de una vez.

Invariante que debes mantener:

```
        < pivote        == pivote        sin revisar        > pivote
   ┌──────────────┬─────────────────┬─────────────────┬──────────────┐
   lo           lt-1  lt          i-1  i             gt  gt+1        hi
```


````python
# TODO: Implementa la partición tricotómica (bandera holandesa)
# REQUISITOS:
# - pivote = arr[lo]
# - lt, i arrancan en lo, lo+1;  gt arranca en hi
# - mientras i <= gt, mira arr[i]:
#     * menor que el pivote  → swap(arr[lt], arr[i]); lt += 1; i += 1
#     * mayor que el pivote  → swap(arr[i], arr[gt]); gt -= 1   (¡i NO avanza!)
#     * igual al pivote      → i += 1
# - Retorna la tupla (lt, gt): arr[lt..gt] son todos iguales al pivote

def particionar_tres_vias(arr, lo, hi):
    """
    Partición tricotómica de Dijkstra.

    Retorna:
        tuple[int, int]: (lt, gt) tal que arr[lt..gt] == pivote y ya están
                         en su posición definitiva
    """
    # TU CÓDIGO AQUÍ
    pass


def quicksort_tres_vias(arr, lo=None, hi=None):
    """
    Quicksort tricotómico.

    Ojo con las llamadas recursivas: toda la zona de iguales queda fija, así
    que la recursión NO la incluye.
        quicksort_tres_vias(arr, lo, lt - 1)
        quicksort_tres_vias(arr, gt + 1, hi)
    """
    if lo is None:
        lo = 0
    if hi is None:
        hi = len(arr) - 1
    # TU CÓDIGO AQUÍ
    pass
````
