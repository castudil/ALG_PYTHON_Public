## PARTE 1D: Implementar BinarySearchST (15 minutos)

Implementa `rank`, `get` y `put` sobre arreglo ordenado.
El esqueleto de la clase ya está dado — completa solo los métodos marcados con `# TU CÓDIGO AQUÍ`.

````python
# TODO: Implementa BinarySearchST
# REQUISITOS:
#   - rank: búsqueda binaria — retorna número de claves < key
#   - get: usa rank para localizar, O(log N)
#   - put: usa rank para localizar; si no existe, desplaza y agrega
#   - min / max: O(1) gracias al arreglo ordenado
#   - keys(): retornar lista de claves en orden

class BinarySearchST:
    """
    Symbol Table con búsqueda binaria sobre arreglo ordenado.
    Complejidad: get O(log N), put O(N).
    """

    def __init__(self, capacity=16):
        self._keys = [None] * capacity
        self._vals = [None] * capacity
        self._n    = 0

    def size(self):
        return self._n

    def isEmpty(self):
        return self._n == 0

    def rank(self, key):
        """
        Retorna el número de claves en la tabla menores que key.
        Implementación: búsqueda binaria sobre self._keys[0..n-1].

        Invariante al terminar:
          - Si key existe:    self._keys[rank(key)] == key
          - Si no existe:     rank(key) es la posición de inserción correcta
        """
        # TU CÓDIGO AQUÍ
        pass

    def get(self, key):
        """
        Retornar el valor asociado a key, o None si no existe — O(log N).
        Lanzar ValueError si key es None.
        HINT: usa rank(key) para localizar, luego verifica que keys[i] == key.
        """
        # TU CÓDIGO AQUÍ
        pass

    def put(self, key, value):
        """
        Insertar o actualizar — O(log N) para ubicar, O(N) para desplazar.
        Lanzar ValueError si key es None.
        Si value es None: llamar a delete(key).
        HINT:
          i = rank(key)
          Si keys[i] == key: actualizar vals[i]
          Si no: desplazar keys[i..n-1] a la derecha e insertar en posición i
        """
        # TU CÓDIGO AQUÍ
        pass

    def contains(self, key):
        return self.get(key) is not None

    def delete(self, key):
        """Eliminar key — O(N) por desplazamiento."""
        if not self.contains(key):
            return
        i = self.rank(key)
        for j in range(i, self._n - 1):
            self._keys[j] = self._keys[j + 1]
            self._vals[j] = self._vals[j + 1]
        self._n -= 1
        self._keys[self._n] = None
        self._vals[self._n] = None

    def min(self):
        """Clave mínima — O(1)."""
        # TU CÓDIGO AQUÍ
        pass

    def max(self):
        """Clave máxima — O(1)."""
        # TU CÓDIGO AQUÍ
        pass

    def keys(self):
        """Lista de claves en orden."""
        return self._keys[:self._n]

    def _resize(self, new_cap):
        new_keys = [None] * new_cap
        new_vals = [None] * new_cap
        for i in range(self._n):
            new_keys[i] = self._keys[i]
            new_vals[i] = self._vals[i]
        self._keys = new_keys
        self._vals = new_vals
````
