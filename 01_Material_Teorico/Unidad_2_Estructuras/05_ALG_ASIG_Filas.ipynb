## Descripción del Problema

### Contexto
Una empresa de soporte técnico recibe tickets de clientes. Cada ticket describe un problema a resolver. Los tickets deben atenderse en orden de llegada (**FIFO**), pero los tickets urgentes tienen **prioridad** y se atienden antes que los normales.

### Estructura del Ticket
Cada ticket es una tupla:
```python
(ticket_id, descripcion, prioridad)
```

Donde:
- **ticket_id** (int): Número único identificador
- **descripcion** (str): Descripción del problema
- **prioridad** (int): 1=normal, 2=urgente

### Operaciones Requeridas
Tu clase `SistemaTickets` debe implementar:

1. **`registrar(ticket_id, descripcion, prioridad)`**  
   - Agrega un nuevo ticket a la fila correspondiente según prioridad
   - Prioridad 2 → fila_urgente
   - Prioridad 1 → fila_normal

2. **`atender()`**  
   - Atiende el siguiente ticket según reglas de prioridad
   - Primero atiende TODOS los urgentes (si hay)
   - Luego atiende los normales
   - Retorna `(ticket_id, descripcion, prioridad)` o `None` si ambas filas vacías

3. **`hay_tickets()`**  
   - Retorna `True` si hay tickets pendientes, `False` si no

4. **`tickets_urgentes()`**  
   - Retorna el número de tickets urgentes pendientes

5. **`tickets_normales()`**  
   - Retorna el número de tickets normales pendientes

6. **`total_pendientes()`**  
   - Retorna la suma de tickets urgentes + normales

7. **`__repr__()`**  
   - Retorna un string descriptivo: `SistemaTickets(urgentes=N, normales=M)`

### Ejemplo de Sesión

```python
sistema = SistemaTickets()

# Registro de tickets
sistema.registrar(1, "no enciende PC", 1)         # normal
sistema.registrar(2, "servidor caído", 2)        # urgente
sistema.registrar(3, "correo lento", 1)          # normal
sistema.registrar(4, "red caída", 2)             # urgente

# Estado: urgentes=[2,4], normales=[1,3]

# Atención de tickets
print(sistema.atender())  # (2, "servidor caído", 2)    [urgente primero]
print(sistema.atender())  # (4, "red caída", 2)         [urgente]
print(sistema.atender())  # (1, "no enciende PC", 1)    [ahora normal]
print(sistema.atender())  # (3, "correo lento", 1)      [normal]
print(sistema.atender())  # None                         [ambas vacías]

print(sistema.hay_tickets())        # False
print(sistema.total_pendientes())   # 0
```

## API de filas.py: Referencia Rápida

La librería `filas.py` proporciona varias implementaciones de filas. En esta tarea usaremos **`FilaDeque`** (basada en `collections.deque`).

| Operación | Descripción | Ejemplo | Complejidad |
|-----------|-------------|---------|-------------|
| `FilaDeque()` | Constructor | `f = FilaDeque()` | O(1) |
| `enqueue(item)` | Agrega al final | `f.enqueue(5)` | O(1) |
| `dequeue()` | Extrae del frente | `x = f.dequeue()` | O(1) |
| `front()` | Ve sin extraer | `x = f.front()` | O(1) |
| `is_empty()` | ¿Vacía? | `if f.is_empty(): ...` | O(1) |
| `size()` | Número elementos | `n = f.size()` | O(1) |
| `len(f)` | Alias de size() | `n = len(f)` | O(1) |
| `repr(f)` | Representación | `print(f)` | O(n) |

### Notas Importantes
- **`dequeue()` y `front()` lanzan `IndexError`** si la fila está vacía → **siempre verificar con `is_empty()` primero**
- Representación visual: `FilaDeque(frente → A → B ← final)`
- En Python, `len(f)` es equivalente a `f.size()`

## Representación Interna: Dos Filas Independientes

La solución usa el patrón de **multiprocesamiento con prioridades**: dos filas completamente independientes.

```python
class SistemaTickets:
    def __init__(self):
        self.__fila_urgente = FilaDeque()   # Prioridad 2
        self.__fila_normal = FilaDeque()    # Prioridad 1
```

### Lógica de Registro
```python
def registrar(ticket_id, descripcion, prioridad):
    ticket = (ticket_id, descripcion, prioridad)
    if prioridad == 2:
        self.__fila_urgente.enqueue(ticket)
    else:
        self.__fila_normal.enqueue(ticket)
```

### Lógica de Atención (CON PRIORIDAD)
```python
def atender():
    # Primero atiende TODOS los urgentes
    if not self.__fila_urgente.is_empty():
        return self.__fila_urgente.dequeue()
    # Luego atiende los normales
    elif not self.__fila_normal.is_empty():
        return self.__fila_normal.dequeue()
    # Si ambas vacías
    else:
        return None
```

### Estado de Ejemplo
Después de los 4 registros del ejemplo:
```
fila_urgente:   [ticket-2, ticket-4]  ← (2, "servidor caído", 2), (4, "red caída", 2)
fila_normal:    [ticket-1, ticket-3]  ← (1, "no enciende PC", 1), (3, "correo lento", 1)
```

Cuando atiendemos:
1. Dequeue urgente → ticket-2
2. Dequeue urgente → ticket-4
3. Dequeue normal → ticket-1
4. Dequeue normal → ticket-3
5. Ambas vacías → None
