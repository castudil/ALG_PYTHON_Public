## Descripción del Problema

### Contexto
Implementa un simulador de mazo de cartas francés (52 cartas: 4 palos × 13 valores).
El simulador debe poder ordenar, barajar y analizar el mazo usando los algoritmos del curso.

### Representación de Cartas
Cada carta es un entero del 1 al 52:
- Cartas 1-13: ♣ Trébol (1=A♣, 11=J♣, 12=Q♣, 13=K♣)
- Cartas 14-26: ♦ Diamante (14=A♦, ..., 26=K♦)
- Cartas 27-39: ♥ Corazón (27=A♥, ..., 39=K♥)
- Cartas 40-52: ♠ Espada (40=A♠, ..., 52=K♠)

El orden "natural" del mazo es [1, 2, 3, ..., 52] (A♣ primero, K♠ al final).

### Operaciones Requeridas

Tu clase `ManejadorMazo` debe implementar:

1. **`cargar_mazo()`** — Carga el mazo en orden natural [1..52]
2. **`barajar()`** — Baraja el mazo con Knuth Shuffle
3. **`ordenar()`** — Ordena el mazo con Selection Sort
4. **`esta_ordenado()`** — Retorna True si el mazo está en orden natural
5. **`tomar_carta()`** — Extrae y retorna la primera carta del mazo (o None si vacío)
6. **`ver_tope()`** — Retorna la primera carta sin extraerla (o None si vacío)
7. **`cartas_restantes()`** — Retorna cuántas cartas quedan
8. **`estadisticas_ultimo_ordenamiento()`** — Retorna dict con comparaciones y swaps del último `ordenar()`
9. **`__repr__()`** — Retorna `ManejadorMazo(cartas=N, ordenado=True/False)`

### Ejemplo de Sesión

```python
m = ManejadorMazo()
m.cargar_mazo()
print(m.esta_ordenado())        # True (mazo recién cargado)
print(m.cartas_restantes())     # 52

m.barajar()
print(m.esta_ordenado())        # False (barajado)

m.ordenar()
print(m.esta_ordenado())        # True (ordenado)

stats = m.estadisticas_ultimo_ordenamiento()
print(stats['comparaciones'])   # 1326  (= 52*51/2)
print(stats['swaps'])           # ≤ 51

carta = m.tomar_carta()
print(carta)                    # 1 (A♣ — primera carta)
print(m.cartas_restantes())     # 51
```

## Suite de Tests Automáticos (NO MODIFICAR)

````python
# NO MODIFICAR ESTA CELDA
import unittest

class TestManejadorMazo(unittest.TestCase):

    def setUp(self):
        self.m = ManejadorMazo()

    def test_01_init_mazo_vacio(self):
        """T01: Mazo recién creado está vacío."""
        self.assertEqual(self.m.cartas_restantes(), 0)

    def test_02_cargar_mazo(self):
        """T02: cargar_mazo() carga 52 cartas en orden."""
        self.m.cargar_mazo()
        self.assertEqual(self.m.cartas_restantes(), 52)
        self.assertTrue(self.m.esta_ordenado())

    def test_03_barajar_conserva_cartas(self):
        """T03: barajar() no pierde cartas."""
        self.m.cargar_mazo()
        self.m.barajar()
        self.assertEqual(self.m.cartas_restantes(), 52)

    def test_04_barajar_desordena(self):
        """T04: barajar() normalmente desordena el mazo."""
        import random
        random.seed(99)
        self.m.cargar_mazo()
        self.m.barajar()
        # Con probabilidad 1/52\! ≈ 0 el mazo queda ordenado
        self.assertFalse(self.m.esta_ordenado())

    def test_05_ordenar_ordena(self):
        """T05: ordenar() deja el mazo en orden natural."""
        self.m.cargar_mazo()
        self.m.barajar()
        self.m.ordenar()
        self.assertTrue(self.m.esta_ordenado())
        self.assertEqual(self.m.cartas_restantes(), 52)

    def test_06_ordenar_estadisticas(self):
        """T06: estadisticas_ultimo_ordenamiento() retorna comp == 52*51/2."""
        self.m.cargar_mazo()
        self.m.ordenar()
        stats = self.m.estadisticas_ultimo_ordenamiento()
        self.assertIn('comparaciones', stats)
        self.assertIn('swaps', stats)
        self.assertEqual(stats['comparaciones'], 52 * 51 // 2)

    def test_07_estadisticas_sin_ordenar(self):
        """T07: estadisticas antes de ordenar retorna 0."""
        self.m.cargar_mazo()
        stats = self.m.estadisticas_ultimo_ordenamiento()
        self.assertEqual(stats['comparaciones'], 0)
        self.assertEqual(stats['swaps'], 0)

    def test_08_tomar_carta_fifo(self):
        """T08: tomar_carta() extrae cartas en orden (desde el tope)."""
        self.m.cargar_mazo()
        carta1 = self.m.tomar_carta()
        carta2 = self.m.tomar_carta()
        self.assertEqual(carta1, 1)
        self.assertEqual(carta2, 2)
        self.assertEqual(self.m.cartas_restantes(), 50)

    def test_09_tomar_mazo_vacio(self):
        """T09: tomar_carta() retorna None si el mazo está vacío."""
        self.m.cargar_mazo()
        for _ in range(52):
            self.m.tomar_carta()
        self.assertIsNone(self.m.tomar_carta())

    def test_10_ver_tope_no_extrae(self):
        """T10: ver_tope() no modifica el mazo."""
        self.m.cargar_mazo()
        tope = self.m.ver_tope()
        self.assertEqual(tope, 1)
        self.assertEqual(self.m.cartas_restantes(), 52)

    def test_11_ver_tope_vacio(self):
        """T11: ver_tope() retorna None si el mazo está vacío."""
        self.assertIsNone(self.m.ver_tope())

    def test_12_ciclo_completo(self):
        """T12: Cargar → barajar → ordenar → tomar todas las cartas."""
        import random
        random.seed(7)
        self.m.cargar_mazo()
        self.m.barajar()
        self.m.ordenar()
        cartas = []
        while self.m.cartas_restantes() > 0:
            cartas.append(self.m.tomar_carta())
        self.assertEqual(cartas, list(range(1, 53)))

    def test_13_repr_str(self):
        """T13: __repr__ retorna string con información útil."""
        self.m.cargar_mazo()
        r = repr(self.m)
        self.assertIn("ManejadorMazo", r)
        self.assertIn("52", r)

if __name__ == '__main__':
    suite = unittest.TestLoader().loadTestsFromTestCase(TestManejadorMazo)
    runner = unittest.TextTestRunner(verbosity=2)
    result = runner.run(suite)
    print()
    print(f"=== RESUMEN === Tests: {result.testsRun} | Exitosos: {result.testsRun - len(result.failures) - len(result.errors)} | Fallos: {len(result.failures)}")
````
