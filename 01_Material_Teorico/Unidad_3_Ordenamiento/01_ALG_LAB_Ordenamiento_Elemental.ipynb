```
...
Paso i=n-2: elegir j aleatorio en [n-2, n-1] → swap(a[n-2], a[j])
```

````python
# NO MODIFICAR ESTA CELDA

def test_mi_knuth_shuffle():
    print("Iniciando tests para MiKnuthShuffle...\n")
    test_count = 0
    passed = 0

    # T1: Resultado tiene los mismos elementos
    test_count += 1
    try:
        k = MiKnuthShuffle(semilla=0)
        lista = [1, 2, 3, 4, 5]
        k.barajar(lista)
        assert sorted(lista) == [1, 2, 3, 4, 5], f"Perdió elementos: {lista}"
        print(f"✓ Test {test_count}: Conserva elementos OK"); passed += 1
    except Exception as e:
        print(f"✗ Test {test_count}: Conserva elementos FALLÓ — {e}")

    # T2: Con semilla fija, resultado es reproducible
    test_count += 1
    try:
        k1 = MiKnuthShuffle(semilla=42)
        lista1 = [1, 2, 3, 4, 5]
        k1.barajar(lista1)

        k2 = MiKnuthShuffle(semilla=42)
        lista2 = [1, 2, 3, 4, 5]
        k2.barajar(lista2)

        assert lista1 == lista2, f"Con misma semilla debería dar mismo resultado"
        print(f"✓ Test {test_count}: Reproducibilidad con semilla OK"); passed += 1
    except Exception as e:
        print(f"✗ Test {test_count}: Reproducibilidad FALLÓ — {e}")

    # T3: swaps() retorna entero no negativo
    test_count += 1
    try:
        k = MiKnuthShuffle()
        k.barajar([1, 2, 3, 4, 5])
        assert isinstance(k.swaps(), int) and k.swaps() >= 0
        assert k.swaps() <= 4  # máximo n-1 swaps para n=5
        print(f"✓ Test {test_count}: swaps() OK (swaps={k.swaps()})"); passed += 1
    except Exception as e:
        print(f"✗ Test {test_count}: swaps() FALLÓ — {e}")

    # T4: Lista de un elemento no cambia
    test_count += 1
    try:
        k = MiKnuthShuffle()
        lista = [99]
        k.barajar(lista)
        assert lista == [99]
        assert k.swaps() == 0
        print(f"✓ Test {test_count}: Un elemento OK"); passed += 1
    except Exception as e:
        print(f"✗ Test {test_count}: Un elemento FALLÓ — {e}")

    # T5: Prueba de uniformidad (estadística)
    test_count += 1
    try:
        N = 30_000
        conteo = Counter()
        for _ in range(N):
            k = MiKnuthShuffle()
            lista = [1, 2, 3]
            k.barajar(lista)
            conteo[tuple(lista)] += 1

        esperado = N / 6  # 6 permutaciones para n=3
        max_desviacion = max(abs(v - esperado) / esperado for v in conteo.values())
        assert len(conteo) == 6, f"Solo {len(conteo)} permutaciones distintas (esperado 6)"
        assert max_desviacion < 0.10, f"Distribución sesgada: desv máx = {max_desviacion:.2%}"
        print(f"✓ Test {test_count}: Uniformidad OK (desv máx = {max_desviacion:.2%})"); passed += 1
    except Exception as e:
        print(f"✗ Test {test_count}: Uniformidad FALLÓ — {e}")

    print(f"\n{'='*50}")
    print(f"Resultado: {passed}/{test_count} tests pasados")
    if passed == test_count:
        print("🎉 ¡Implementación correcta\!")
    else:
        print(f"⚠️  {test_count - passed} test(s) fallido(s).")
    print(f"{'='*50}")

test_mi_knuth_shuffle()
````

## PARTE 4: Benchmark (5 minutos)

Compara el rendimiento de tu implementación vs la del módulo.


````python
# NO MODIFICAR ESTA CELDA
import time

def benchmark(n=2000):
    print(f"Benchmark con n={n:,} elementos\n{'='*60}")

    # Módulo vs tu implementación
    for nombre, sorter in [('Módulo SelectionSort', SelectionSort()),
                            ('MiSelectionSort', MiSelectionSort())]:
        lista = list(range(n, 0, -1))  # peor caso: orden inverso
        t0 = time.perf_counter()
        sorter.ordenar(lista)
        t1 = time.perf_counter()
        print(f"{nombre:<25} → {(t1-t0)*1000:.1f}ms | comp={sorter.comparaciones():,}")

    print()
    for nombre, shuffler in [('Módulo KnuthShuffle', KnuthShuffle()),
                              ('MiKnuthShuffle', MiKnuthShuffle())]:
        lista = list(range(n))
        t0 = time.perf_counter()
        shuffler.barajar(lista)
        t1 = time.perf_counter()
        print(f"{nombre:<25} → {(t1-t0)*1000:.1f}ms | swaps={shuffler.swaps():,}")

benchmark()
````
