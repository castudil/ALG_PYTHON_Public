```python
# Setup del laboratorio — ejecutar primero
import sys, random, time, timeit
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

print("✅ Imports correctos")
print("🐍 Python", sys.version.split()[0])

# ── Implementaciones de referencia (usadas en los verificadores) ──────────

def _insertion_ref(a):
    a = a[:]
    for i in range(1, len(a)):
        c = a[i]; j = i - 1
        while j >= 0 and a[j] > c: a[j+1] = a[j]; j -= 1
        a[j+1] = c
    return a

def _bubble_ref(a):
    a = a[:]
    for i in range(len(a)-1):
        hubo = False
        for j in range(len(a)-1-i):
            if a[j] > a[j+1]: a[j], a[j+1] = a[j+1], a[j]; hubo = True
        if not hubo: break
    return a

def _shell_ref(a, gaps=None):
    a = a[:]
    n = len(a)
    if gaps is None:
        gaps = []
        g = 1
        while g < n//3: g = 3*g+1; gaps.append(g)
        gaps = sorted(gaps, reverse=True) or [1]
    for gap in gaps:
        for i in range(gap, n):
            c = a[i]; j = i - gap
            while j >= 0 and a[j] > c: a[j+gap] = a[j]; j -= gap
            a[j+gap] = c
    return a

def _counting_ref(a, k=None):
    if not a: return []
    if k is None: k = max(a) + 1
    cnt = [0]*k
    for v in a: cnt[v] += 1
    res = []
    for v in range(k): res.extend([v]*cnt[v])
    return res

print("✅ Implementaciones de referencia cargadas")
```

## PARTE 2C: Benchmark — Shell Sort vs Insertion Sort (10 minutos)

```python
# Benchmark: Shell Sort vs Insertion Sort — curva de escalabilidad
import timeit, random
import matplotlib.pyplot as plt
import numpy as np

def insertion_simple(a):
    a = a[:]
    for i in range(1, len(a)):
        c = a[i]; j = i - 1
        while j >= 0 and a[j] > c: a[j+1] = a[j]; j -= 1
        a[j+1] = c
    return a

def shell_knuth(a):
    a = a[:]
    n = len(a)
    gaps = []
    g = 1
    while g < n//3: g = 3*g+1; gaps.append(g)
    gaps = sorted(gaps, reverse=True) or [1]
    for gap in gaps:
        for i in range(gap, n):
            c = a[i]; j = i - gap
            while j >= 0 and a[j] > c: a[j+gap] = a[j]; j -= gap
            a[j+gap] = c
    return a

ns = [100, 500, 1000, 2000, 5000, 10000]
t_ins, t_shell = [], []
for n in ns:
    datos = random.sample(range(n*3), n)
    t_ins.append(timeit.timeit(lambda: insertion_simple(datos), number=15) / 15 * 1000)
    t_shell.append(timeit.timeit(lambda: shell_knuth(datos),    number=15) / 15 * 1000)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4))

# Tiempo absoluto
ax1.plot(ns, t_ins,   'o-', color='#2196F3', label='Insertion Sort', linewidth=2)
ax1.plot(ns, t_shell, 's-', color='#9C27B0', label='Shell Sort (Knuth)', linewidth=2)
ax1.set_xlabel('n'); ax1.set_ylabel('Tiempo (ms)')
ax1.set_title('Tiempo absoluto'); ax1.legend(); ax1.grid(alpha=0.3)

# Speedup de Shell sobre Insertion
speedup = [i/s for i, s in zip(t_ins, t_shell)]
ax2.bar(range(len(ns)), speedup, color='#9C27B0', alpha=0.8)
ax2.set_xticks(range(len(ns))); ax2.set_xticklabels(ns, rotation=30)
ax2.set_xlabel('n'); ax2.set_ylabel('Speedup (veces más rápido)')
ax2.set_title('Speedup de Shell Sort vs Insertion Sort')
ax2.axhline(y=1, color='red', linestyle='--', alpha=0.5)
ax2.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n{'n':>6} | {'Insertion':>10} | {'Shell':>10} | {'Speedup':>8}")
print("-" * 40)
for n, ti, ts, sp in zip(ns, t_ins, t_shell, speedup):
    print(f"{n:>6} | {ti:>8.2f}ms | {ts:>8.2f}ms | {sp:>7.1f}×")
```

## PARTE 2E: Tabla Comparativa Final — Los 5 Algoritmos (10 minutos)

```python
# Benchmark final: los 5 algoritmos lado a lado
import timeit, random
import matplotlib.pyplot as plt

def counting_simple(a):
    if not a: return []
    k = max(a)+1; cnt=[0]*k
    for v in a: cnt[v]+=1
    res=[]
    for v in range(k): res.extend([v]*cnt[v])
    return res

def selection_sort(a):
    a = a[:]
    n = len(a)
    for i in range(n-1):
        idx = i
        for j in range(i+1, n):
            if a[j] < a[idx]: idx = j
        a[i], a[idx] = a[idx], a[i]
    return a

algoritmos = {
    'Selection': selection_sort,
    'Insertion': _insertion_ref,
    'Bubble':    _bubble_ref,
    'Shell':     _shell_ref,
    'Counting':  counting_simple,
}

ns = [100, 500, 1000, 3000]
colores = ['#F44336', '#2196F3', '#FF9800', '#9C27B0', '#4CAF50']
fig, ax = plt.subplots(figsize=(11, 5))

for (nombre, fn), color in zip(algoritmos.items(), colores):
    tiempos = []
    for n in ns:
        datos = random.sample(range(n*3), n)
        t = timeit.timeit(lambda: fn(datos[:]), number=20) / 20 * 1000
        tiempos.append(t)
    ax.plot(ns, tiempos, 'o-', color=color, label=nombre, linewidth=2)

ax.set_xlabel('n'); ax.set_ylabel('Tiempo (ms)')
ax.set_title('Los 5 algoritmos de la semana — Comparación de tiempos')
ax.legend(loc='upper left'); ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
